```python
# Import necessary libraries for TensorFlow and NumPy.
import tensorflow as tf
import numpy as np
```

```python
# Define input features (X) and target labels (y) for an AND gate.
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
], dtype=np.float32)

y = np.array([
    [0],
    [0],
    [0],
    [1]
], dtype=np.float32)


print(X)
print(y)
```

```python
# Define a simple sequential Keras model with a single dense layer and sigmoid activation for binary classification.
model = tf.keras.Sequential([
    tf.keras.layers.Dense(
        units=1,
        activation='sigmoid',
        input_shape=(2,)
    )
])
```

```python
# Compile the model using SGD optimizer and binary cross-entropy loss.
model.compile(
    optimizer='sgd',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
# Train the model on the defined dataset for 500 epochs.
model.fit(
    X,
    y,
    epochs=500,
    verbose=1
)
```

```python
# Make predictions on the input data using the trained model.
prediction = model.predict(X)

print("Predicted Values")

for i in range(len(X)):
    print(X[i], "->", prediction[i][0])
```

```python
# Convert the probabilistic predictions into binary outputs (0 or 1) using a 0.5 threshold.
binary_output = (prediction > 0.5).astype(int)

print(binary_output)
```

```python
# Retrieve and display the learned weights and bias of the model's dense layer.
weights, bias = model.layers[0].get_weights()

print("Weights")

print(weights)

print("\nBias")

print(bias)
```
